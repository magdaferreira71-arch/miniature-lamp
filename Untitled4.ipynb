```python
# lista de notas
notas= (7.0,8.0,9,0, 10.0)

# calculo de media
media= sum (notas) / len (notas)

# Determinação da Situação
if media >= 7.0:
        situacao = "Aprovado"
else:
        situacao = "Reprovado"

    # Relatório Final

print("Notas:", notas)
print("Média:", media)
print("Situação:", situacao)
```
