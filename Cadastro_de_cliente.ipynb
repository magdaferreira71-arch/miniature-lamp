```python
import sqlite3
import ipywidgets as widgets
from IPython.display import display

# Função para conectar e criar o banco de dados
def conectar_banco():
    conexao = sqlite3.connect("clientes.db")
    cursor = conexao.cursor()
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS clientes (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            nome TEXT,
            telefone TEXT,
            mensagem TEXT
        )
    """)
    conexao.commit()
    conexao.close()

# Inicializar o banco
conectar_banco()

# Criar os componentes da interface com ipywidgets
entry_nome = widgets.Text(description="Nome:", placeholder="Nome do Cliente")
entry_telefone = widgets.Text(description="Telefone:", placeholder="Telefone/WhatsApp")
entry_msg = widgets.Text(description="Mensagem:", placeholder="Motivo do contato")
botao_salvar = widgets.Button(description="Salvar Atendimento", button_style="success")
out_feedback = widgets.Output()

# Função para salvar o atendimento/cliente
def cadastrar(b):
    nome = entry_nome.value.strip()
    telefone = entry_telefone.value.strip()
    mensagem = entry_msg.value.strip()

    with out_feedback:
        out_feedback.clear_output()
        if not nome or not telefone:
            print("⚠️ Aviso: Preencha nome e telefone!")
            return

        conexao = sqlite3.connect("clientes.db")
        cursor = conexao.cursor()
        cursor.execute("INSERT INTO clientes (nome, telefone, mensagem) VALUES (?, ?, ?)", (nome, telefone, mensagem))
        conexao.commit()
        conexao.close()

        print("✅ Sucesso: Cliente cadastrado com sucesso!")
        entry_nome.value = ""
        entry_telefone.value = ""
        entry_msg.value = ""

# Associar evento ao botão
botao_salvar.on_click(cadastrar)

# Exibir a interface
print("=== Ferramenta de Atendimento ao Cliente ===")
display(entry_nome, entry_telefone, entry_msg, botao_salvar, out_feedback)
```
